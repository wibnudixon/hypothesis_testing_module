## 4. Analysis

Derived metrics and aggregations built from the data above.

```python
# --- Key match summary stats ---
print("=== Key Match Stats — Full Detail ===")
display_cols = [
    "date", "home_team", "away_team",
    "home_goals", "away_goals", "result",
    "home_xg", "away_xg", "xg_diff",
    "home_possession", "away_possession",
    "home_shots_on_target", "away_shots_on_target",
]
print(key_matches[display_cols].to_string(index=False))
```

```python
# --- Head-to-head aggregate record ---
def h2h_record(df, team):
    """Return wins/draws/losses for a given team from the H2H table."""
    wins   = (df["winner"] == team).sum()
    draws  = (df["winner"] == "Draw").sum()
    losses = len(df) - wins - draws
    gf     = df.loc[df["home"] == team, "hg"].sum() + df.loc[df["away"] == team, "ag"].sum()
    ga     = df.loc[df["home"] == team, "ag"].sum() + df.loc[df["away"] == team, "hg"].sum()
    return {"team": team, "wins": wins, "draws": draws, "losses": losses, "gf": gf, "ga": ga, "gd": gf - ga}

teams = ["Arsenal", "Man City"]
h2h_summary = pd.DataFrame([h2h_record(h2h, t) for t in teams])

print("=== Head-to-Head Aggregate (last 6 PL seasons) ===")
print(h2h_summary.to_string(index=False))
```

```python
# --- Season trend: points over time ---
print("=== Points Trend by Season ===")
pivot_pts = seasons.pivot(index="season", columns="team", values="points")
print(pivot_pts)
```

```python
# --- xG over/under-performance in key matches ---
key_matches["home_xg_diff"] = key_matches["home_goals"] - key_matches["home_xg"]
key_matches["away_xg_diff"] = key_matches["away_goals"] - key_matches["away_xg"]

print("=== xG Over/Under-performance (key matches) ===")
print(
    key_matches[["date", "home_team", "away_team", "home_xg_diff", "away_xg_diff"]]
    .to_string(index=False)
)
```

## 5. Visualisations

```python
# --- Chart 1: Points trajectory ---
fig, ax = plt.subplots()
for team, grp in seasons.groupby("team"):
    ax.plot(grp["season"], grp["points"], marker="o", linewidth=2, label=team)

ax.set_title("Premier League Points Trajectory (2019/20 – 2023/24)")
ax.set_xlabel("Season")
ax.set_ylabel("Points")
ax.yaxis.set_major_locator(mticker.MultipleLocator(5))
ax.legend()
plt.tight_layout()
plt.savefig("points_trajectory.png", dpi=150)
plt.show()
```

```python
# --- Chart 2: xG vs Goals in key matches ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, side, label in zip(axes, ["home", "away"], ["Home Team", "Away Team"]):
    x = np.arange(len(key_matches))
    width = 0.35
    labels = [
        f"{r[f'{side}_team']}\n{r['date'].strftime('%b %Y')}"
        for _, r in key_matches.iterrows()
    ]
    ax.bar(x - width / 2, key_matches[f"{side}_goals"], width, label="Goals", color="steelblue")
    ax.bar(x + width / 2, key_matches[f"{side}_xg"],    width, label="xG",    color="salmon", alpha=0.8)
    ax.set_xticks(x)
    ax.set_xticklabels(labels, fontsize=8)
    ax.set_title(f"{label}: Goals vs xG (key matches)")
    ax.set_ylabel("Value")
    ax.legend()

plt.suptitle("Goals vs Expected Goals — Key Matches", fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig("xg_vs_goals.png", dpi=150)
plt.show()
```

```python
# --- Chart 3: Head-to-head win/draw/loss breakdown ---
h2h_counts = h2h["winner"].value_counts()

fig, ax = plt.subplots(figsize=(7, 5))
colours = {"Arsenal": "#EF0107", "Man City": "#6CABDD", "Draw": "#888888"}
bar_colours = [colours.get(k, "grey") for k in h2h_counts.index]
h2h_counts.plot(kind="bar", ax=ax, color=bar_colours, edgecolor="white", width=0.5)

ax.set_title("Head-to-Head Results — Arsenal vs Man City (Last 6 PL Seasons)")
ax.set_xlabel("")
ax.set_ylabel("Number of Matches")
ax.set_xticklabels(h2h_counts.index, rotation=0)
ax.yaxis.set_major_locator(mticker.MultipleLocator(1))
plt.tight_layout()
plt.savefig("h2h_results.png", dpi=150)
plt.show()
```
